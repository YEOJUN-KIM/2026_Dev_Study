### Series

- 1차원 데이터를 표현하는 자료구조. `인덱스`가 붙어있는 한줄짜리 데이터. 1차원 배열과 흡사

```python
np_scores = np.array([85, 90, 95, 80])

np_scores
```

```python
# 기본 인덱스를 사용하는 Series 생성과 값 조회
scores = pd.Series([85, 90, 95, 80])

display(scores)
display(Markdown(f"**첫 번째 점수:** {scores.iloc[0]}"))
```

#### Series 인덱스 지정

- Series 생성 시 인덱스를 원하는 값으로 지정할 수 있음

```python
# 이름을 인덱스로 사용하는 Series 생성과 값 조회
scores = pd.Series(
    [85, 90, 95, 80],
    index=['홍길동', '김철수', '이영희', '박민수']
)

display(scores)
display(Markdown(f"**홍길동 점수:** {scores.loc['홍길동']}"))
```

#### Series 연산

```python
# Series 전체 연산과 기본 통계
display(Markdown("**5점씩 더하기:**"))
display(scores + 5)
display(Markdown("**점수를 2로 나누기:**"))
display(scores / 2)
display(Markdown(f"**평균 점수:** {scores.mean()}"))
display(Markdown(f"**최고 점수:** {scores.max()}"))
display(Markdown(f"**데이터 개수:** {scores.count()}"))
```

- NumPy와 통계함수가 거의 유사하게 존재. Pandas만 가지고도 가능
- 속도문제로 예전에는 NumPy를 많이 사용했지만, 현재는 사용효율 높지 않음

### DataFrame

- Pandas 2차원 표형태 데이터 표현하는 자료구조
- Pandas의 기본 자료구조형
- 엑셀, CSV(엑셀형태 텍스트파일), DB 테이블과 거의 동일한 구조

```python
# 파이썬 딕셔너리로 DataFrame 생성
raw_data = {
    'name': ['홍길동', '김철수', '이영희', '박민수', '성유고', '애슐리'],
    'age': [20, 21, 22, 23, 30, 29],
    'major': ['데이터분석', '웹개발', '인공지능', '임베디드', '인공지능', '웹개발'],
    'score': [85, 90, 95, 80, 90, 100]
}

# DataFrame은 일반적으로 df라는 변수명을 사용
df = pd.DataFrame(raw_data)
df
```

#### DataFrame 구조 확인
- NumPy 의 shape, size, dtype, ndim 과 유사한 기능

```python
# 앞쪽과 뒤쪽 데이터 확인
display(Markdown("**앞쪽 3행:**"))
display(df.head(3))
display(Markdown("**뒤쪽 2행:**"))
display(df.tail(2))
```

```python
# DataFrame 구조와 결측값 확인
display(Markdown(f"**행과 열 개수:** {df.shape}"))
display(Markdown(f"**컬럼:** {list(df.columns)}"))
display(Markdown(f"**인덱스:** {list(df.index)}"))
display(Markdown("**상세 정보:**"))
df.info()
```

#### 기본 통계

- 수치타입만 나옴

```python
# 전체 기본 통계
df.describe()
```

```python
# age 컬럼의 주요 통계
display(Markdown(f"**평균:** {df['age'].mean()}"))
display(Markdown(f"**중앙값:** {df['age'].median()}"))
display(Markdown(f"**1사분위수(25%):** {df['age'].quantile(0.25)}"))
display(Markdown(f"**2사분위수(50%):** {df['age'].quantile(0.5)}"))
```

#### `loc`와 `iloc`로 행과 열 선택

- `loc`: 인덱스 이름을 기준으로 선택
- `iloc`: 행과 열의 위치를 기준으로 선택

```python
# loc를 이용한 행과 열 선택
df.loc[0]
# 시리즈의 인덱스로 각 열의 값을 확인
df.loc[0]['name']
```

```python
# 슬라이싱으로 여러행을 선택가능
# Python 슬라이싱 규칙에서 벗어남 -> 뒤에 수가 n-1이 아닌 n까지 나옴
# ex) 0~1이 아닌 0~2까지 나옴 
display(df.loc[:2])   # loc : 인덱스행
# Python과 같은 슬라이싱 기능
display(df.iloc[:2])  # iloc : 위치값행
```

#### 인덱스 변경과 범위 선택

- DataFrame을 복사한 뒤 문자열 인덱스를 지정하고 원하는 범위를 선택

```python
# 기존 df를 복사해서 새로운 DataFrame 생성
df2 = df.copy()

# 문자열 인덱스 지정
df2.index = ['50001', '50002', '50003', '50004', '50005', '50006']

df2
```

```python
df3 = df2.loc['50001':'50003', ['name', 'age']]
df3  # 새로운 DataFrame 확인
```

#### 조건 연산(Where절)

```python
# True / False 판별
display(df['score'] >= 90)

# 조건에 해당하는 내용 출력
display(df[df['score'] >= 90])
```

- NumPy와 동일한 조건 연산
- Python : and,or,not 사용
- NumPy와 Pandas : &, | ~ 사용

```python
# 점수가 90점 이상이며 나이가 21살 이상인 학생을 출력
display(df[(df['score'] >= 90) & (df['age'] >= 21)])

# 점수가 90점 이상이거나 나이가 21살 이상인 학생을 출력
display(df[(df['score'] >= 90) | (df['age'] >= 21)])

# 점수가 90점 미만이고 나이가 21살 이상인 학생을 출력
display(df[(df['score'] < 90) & (df['age'] >= 21)])
```

#### 컬럼 추가

- 기존 DataFrame에 필요한 컬럼이 생길 때 추가

```python
df['pass'] = df['score'] >= 90
df['score_bonus'] = df['score'] + 5
df
```

####  컬럼 수정

```python
# 보너스 점수에 2를 뺀다
df['score_bonus'] = df['score_bonus'] -2
display(df)
# 학생의 점수가 0~100을 벗어날때 정상으로 바꾼다
df.loc[df['score_bonus'] > 100, 'score_bonus'] = 100
df.loc[df['score_bonus'] < 0, 'score_bonus'] = 0
display(df)
```

#### 컬럼 삭제

```python
df.drop(columns=['score_bonus'], inplace=True) # inplace=True는 원본 DataFrame을 직접 변경한다 기본은 False임
df
```

#### 행 추가

```python
df.loc[6] = ['최지우',24,'웹개발',88,True]
df.loc[df.index.max() + 1] = ['박선호',25,'인공지능',91, None]
df
```
