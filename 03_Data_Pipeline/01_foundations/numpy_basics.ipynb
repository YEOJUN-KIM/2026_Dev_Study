#### 결측값 핸들링

- 실제 데이터에서 값이 비어있는 경우, 이를 결측치/결측값이라고 함
- 일반적으로 `NULL`, 파이썬에서는 `None`, NumPy에서는 `np.nan` 로 표현

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 21.8])

# 일반 통계 함수는 결측값이 있으면 nan을 반환
print("원본 데이터:", temperatures)
print("일반 평균:", temperatures.mean())
print("일반 최솟값:", temperatures.min())
print("결측값을 제외한 평균:", np.nanmean(temperatures))
````

````python
# 결측값 위치 확인 후 결측값 제거
missing_mask = np.isnan(temperatures)
valid_temperatures = temperatures[~missing_mask]

print("결측값 위치:", missing_mask)
print("결측값 제거 결과:", valid_temperatures)
````

- 결측치를 정리방법 : 결측치를 제거하거나(건수가 줄어듬), 나머지 평균값을 임의로 기입하거나, 기본통계에서 도출할 수 있는 다른 값들로 채우기

#### 이상치 핸들링

- 일반적인 범위를 벗어나는 비정상 데이터
- 예 부산 1월 일별 평균기온에 -90도. 습도 99%

````python
temperatures = np.array([22.5, 23.0, np.nan, 25.1, 999.0, 21.8, -99.0])

# 결측값을 제거하고 15도 이상 50도 이하인 값만 선택
valid_temps = temperatures[
    (~np.isnan(temperatures)) &
    (temperatures >= 15.0) &
    (temperatures <= 50.0)
]

print("원본 데이터:", temperatures)
print("정상 데이터:", valid_temps)
````

- 전처리(PreProcessing) - 분석 이전에 이상치, 결측치를 없애는 작업

### 실습

#### 기온데이터 전처리

````python
temperatures = np.array([
    22.5, 23.0, np.nan, 25.1, 999.0,
    21.8, -99.0, 24.0, 23.5, np.nan
])

# 결측값과 이상치 제거
missing_count = np.isnan(temperatures).sum()
valid_temps = temperatures[
    (~np.isnan(temperatures)) &
    (temperatures >= -15.0) &
    (temperatures <= 50.0)
]

print("결측값 개수:", missing_count)
print("정상 데이터:", valid_temps)
print("평균 기온:", valid_temps.mean())
print("최저 기온:", valid_temps.min())
print("최고 기온:", valid_temps.max())
````

#### 매출 데이터 전처리

- 결측값 개수 출력
- 결측값 제외
- 0원 10만원 내의 정상 매출만 남김
- 정상 매출의 평균, 최대값, 최소값 출력
- 정상 매출에서 평균이상인 값만 출력

````python
# NumPy 숫자 배열에서는 None 대신 np.nan으로 결측값 표현
raw_sales = [
    12000, 15000, np.nan, 30000, -5000,
    25000, 9999999, 18000, np.nan, 22000
]
sales = np.array(raw_sales)

# 결측값을 제외하고 0원 이상 10만 원 이하의 정상 매출만 선택
missing_count = np.isnan(sales).sum()
valid_sales = sales[
    (~np.isnan(sales)) &
    (sales >= 0) &
    (sales <= 100000)
]

print("결측값 개수:", missing_count)
print("정상 매출:", valid_sales)
````

````python
# 정상 매출 통계와 평균 이상 매출 확인
mean_sales = valid_sales.mean()

print("평균 매출:", mean_sales)
print("최소 매출:", valid_sales.min())
print("최대 매출:", valid_sales.max())
print("평균 이상 매출:", valid_sales[valid_sales >= mean_sales])
````
