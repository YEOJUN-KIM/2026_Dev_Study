# 00. TensorFlow 실행 환경 확인
README의 Ubuntu/WSL + uv 설정 후 실행합니다. 모든 노트는 폴더 루트에서 새 커널로 위에서 아래로 실행합니다.
GPU가 없어도 이후 분류 실습은 CPU에서 실행할 수 있습니다. 메모리 증가 설정은 GPU 연산 전에 적용합니다.

```python
import sys
import tensorflow as tf
from common.runtime import configure_runtime

gpus = configure_runtime()
print('Python:', sys.executable)
print('TensorFlow:', tf.__version__)
print('인식된 물리 장치:', tf.config.list_physical_devices())
print('GPU 사용 가능:', bool(gpus))
```

## 실제 연산 장치 확인
목록에 GPU가 표시되는지와 연산이 GPU에 배치되는지를 각각 확인합니다.
`/device:GPU:0`가 출력되면 아래 행렬 곱셈이 GPU에서 실행된 것입니다.
GPU가 없으면 CPU 결과를 출력합니다. 메모리 증가 옵션은 메모리 사용 상한을 정하는 기능은 아닙니다.

```python
device = '/GPU:0' if gpus else '/CPU:0'
with tf.device(device):
    a = tf.ones((128, 128))
    result = tf.matmul(a, a)
print('연산 장치:', result.device)
print('결과:', result[0, 0].numpy())
```

## 확인할 점
- GPU 목록이 비어 있으면 README의 진단 순서대로 WSL, 드라이버, 커널을 확인합니다.
- 메모리 설정 오류가 나면 커널을 재시작합니다.
- 물리 장치는 공개 API로 확인합니다. 원본의 내부 모듈 `tensorflow.python` 의존과 중복 출력을 제거했습니다.
