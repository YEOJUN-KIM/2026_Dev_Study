## 선형·다항 회귀 기초


### 선형회귀모델(Linear Regression Model)

```python
"""
<특징>
 - 과거에 가장 널리 사용되었던 대표적인 회귀모델
 - 비교적 간단하며, 성능이 뛰어나기 때문에 회귀모델 수행 시 가장 처음에 비교대상 모델로 좋음
 - 선형회귀모델은 하이퍼파라메터 튜닝 속성(변수)이 없음
   -> 튜닝을 할려면 : 독립변수의 선형 분포에 따라 특성을 늘리는(추가하는) 처리를 하면 됩니다.

<선형회귀모델(LinearRegression) 종류>
    - 선형회귀 모델(직선) : 독립변수의 특성(컬럼)이 한개인 경우 사용
    - 다항회귀 모델(곡선) : 독립변수의 특성(컬럼)이 두개인 경우 사용
    - 다중회귀 모델(포물선) : 독립변수의 특성(컬럼)이 여러개인 경우 사용

<사용 라이브러리>
 - 패키지 : sklearn.linear_model
 - 모델(클래스) : LinearRegression
               : 선형/다항/다중회귀 알고리즘(산술식)을 모두 포함하고 있는 모델
 - 모델이 자체적으로 독립변수의 특성 갯수를 확인해서
   직선, 곡선, 포물선 알고리즘(산술식)을 알아서 선택 처리합니다.
"""
```

```python
"""
<전체 분석 순서>
 데이터 수집 > 데이터 전처리 > 가공(데이터 스케일링, 특성 추가, 훈련:검증:테스트 분류 등등)
 
 > 모델 클래스 생성 > 훈련시키기(fit) > 정확도(score)(회귀:결정계수라고 칭함) > 과적합 여부 해석
 > 하이퍼파라메터 튜닝(튜닝이 필요한 경우 수행) > 예측(predict) > 평가(오차:회귀, 지표:분류)
 
 ** 위의 과정을 다른 여러 모델로도 적용
 > 최종 모델 선정을 위힌 모델간 비교 > 최종 선정 모델 파일로 저장 > 서비스 영역에 배포(전달)
"""
```

### 사용할 라이브러리 정의

```python
### 숫자 데이터 처리용
import numpy as np

### 훈련:검증:테스트 데이터 분리용 함수
from sklearn.model_selection import train_test_split
```

### 데이터 정의하기

```python
### 농어의 길이
perch_length= np.array(
    [8.4, 13.7, 15.0, 16.2, 17.4, 18.0, 18.7, 19.0, 19.6, 20.0,
     21.0, 21.0, 21.0, 21.3, 22.0, 22.0, 22.0, 22.0, 22.0, 22.5,
     22.5, 22.7, 23.0, 23.5, 24.0, 24.0, 24.6, 25.0, 25.6, 26.5,
     27.3, 27.5, 27.5, 27.5, 28.0, 28.7, 30.0, 32.8, 34.5, 35.0,
     36.5, 36.0, 37.0, 37.0, 39.0, 39.0, 39.0, 40.0, 40.0, 40.0,
     40.0, 42.0, 43.0, 43.0, 43.5, 44.0]
     )

### 농어 무게
perch_weight = np.array(
    [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
     110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
     130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
     197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
     514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
     820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
     1000.0, 1000.0]
     )

### 각 데이터의 차원 확인하기
print(perch_length.shape, perch_weight.shape)
```

### 독립변수와 종속변수 정의하기

```python
"""
* 주제 : 길이를 이용하여 무게 예측하기
 - 독립변수(X) : 길이
 - 종속변수(y) : 무게
"""
```

### 데이터 분리하기

```python
### 훈련 : 테스트 = 75% : 25%로 분리해주세요.
# 훈련 데이터 변수명 : train_input, train_target
# 테스트 데이터 변수명 : test_input, test_target
# 각 변수들의 차원 확인
train_input, test_input, train_target, test_target = train_test_split(
    perch_length, perch_weight, test_size=0.25, random_state=42
)

print(train_input.shape, train_target.shape)
print(test_input.shape, test_target.shape)
```

### 훈련을 시키기 위한 독립변수의 차원 확인

```python
### 훈련을 위한 독립변수의 차원은 2차원, 종속변수는 1차원
# 독립변수를 2차원으로 수정하기
train_input = train_input.reshape(-1, 1)
test_input  = test_input.reshape(-1, 1)

### 차원 확인하기
print(train_input.shape, train_target.shape)
print(test_input.shape, test_target.shape)
```

### 훈련(학습) 시키기

```python
from sklearn.linear_model import LinearRegression

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
### 1. 훈련에 사용할 모델 클래스 생성시키기
# <사용 라이브러리>
#  - 패키지 : sklearn.linear_model
#  - 모델(클래스) : LinearRegression
lr = LinearRegression()
lr

### 2. 훈련 시키기(fit)
lr.fit(train_input, train_target)

### 3. 훈련 및 테스트의 정확도(결정계수) 확인하기
train_score = lr.score(train_input, train_target)
test_score = lr.score(test_input, test_target)
print(train_score, test_score, train_score-test_score)

### (과적합 해석)
# - 훈련정확도가 테스트정확도보다 0.115의 차이로 과대적합이 발생하고 있음

### 4. 예측하기(테스트 데이터로 예측)
train_pred = lr.predict(train_input)
test_pred = lr.predict(test_input)
print(test_pred)

### 5. 오차 확인하기(훈련 및 테스트 오차 모두 확인)
# 평균절대오차 (0에 가까울수록 좋음)
train_mae = mean_absolute_error(train_target, train_pred)
test_mae = mean_absolute_error(test_target, test_pred)

# 평균제곱오차 (0에 가까울수록 좋음)
train_mse = mean_squared_error(train_target, train_pred)
test_mse = mean_squared_error(test_target, test_pred)

# 결정계수(정확도) (1에 가까울수록 좋음)
train_r2 = r2_score(train_target, train_pred)
test_r2 = r2_score(test_target, test_pred)

print(train_mae, train_mse, train_r2)
print(test_mae, test_mse, test_r2)


### 6. 임의 값(길이 50)으로 예측해 보기
pred_50 = lr.predict([[50]])
print(f"길이 50을 예측한 결과는 {pred_50} 입니다.")
### KNN : 1033.333, 0.9804899950518966 0.9746459963987609 0.005843998653135674
### LR-특성1개 : 1241.839 0.9398463339976041 0.824750312331356 0.1150960216662481
```

### 다항회귀모델 적용하기(곡선의 방정식 적용)

```python
"""
<다항회귀모델>
 - 다항식(곡선의 방정식)을 사용하는 회귀모델
 - 산점도의 분포 형태가 곡선을 띄는 경우에 사용되는 모델 입니다.
 - 특성이 2개 사용됨 : x^2, x
 - 다항식(y) = ax^2 + bx + c
 - a와 b : 기울기(머신러닝에서는 기울기=가중치=계수라고 칭합니다.)
 - c : y절편
 - 다항회귀에서는 x값을 제곱한 특성값이 하나더 추가됩니다.(만들어 줘야 합니다.)
"""
```

### 훈련 및 테스트(검증데이터가 있는 경우 포함) 데이터에 대한 특성 추가하기

```python
### 리스트 구조 : [[1, 2], [3, 4].....]
# - column_stack() : 행열 차원에서 열 추가하는 함수
train_poly = np.column_stack((train_input**2, train_input))
test_poly  = np.column_stack((test_input**2, test_input))

print(train_poly.shape, train_target.shape, test_poly.shape, test_target.shape)
```

### 다항회귀모델 훈련(학습) 하기

```python
### 모델 생성부터 임의의값 50길이로 예측까지 해주세요

### 1. 훈련에 사용할 모델 클래스 생성시키기
# <사용 라이브러리>
#  - 패키지 : sklearn.linear_model
#  - 모델(클래스) : LinearRegression
lr = LinearRegression()
lr

### 2. 훈련 시키기(fit)
lr.fit(train_poly, train_target)

### 3. 훈련 및 테스트의 정확도(결정계수) 확인하기
train_score = lr.score(train_poly, train_target)
test_score = lr.score(test_poly, test_target)
print(train_score, test_score, train_score-test_score)

### (과적합 해석)
# - 훈련정확도가 테스트정확도보다 0.115의 차이로 과대적합이 발생하고 있음

### 4. 예측하기(테스트 데이터로 예측)
train_pred = lr.predict(train_poly)
test_pred = lr.predict(test_poly)
print(test_pred)

### 5. 오차 확인하기(훈련 및 테스트 오차 모두 확인)
# 평균절대오차 (0에 가까울수록 좋음)
train_mae = mean_absolute_error(train_target, train_pred)
test_mae = mean_absolute_error(test_target, test_pred)

# 평균제곱오차 (0에 가까울수록 좋음)
train_mse = mean_squared_error(train_target, train_pred)
test_mse = mean_squared_error(test_target, test_pred)

# 결정계수(정확도) (1에 가까울수록 좋음)
train_r2 = r2_score(train_target, train_pred)
test_r2 = r2_score(test_target, test_pred)

print(train_mae, train_mse, train_r2)
print(test_mae, test_mse, test_r2)


### 6. 임의 값(길이 50)으로 예측해 보기
pred_50 = lr.predict([[50**2, 50]])
print(f"길이 50을 예측한 결과는 {pred_50} 입니다.")
### KNN : 1033.333, 0.9804899950518966 0.9746459963987609 0.005843998653135674
### LR-특성1개 : 1241.839 0.9398463339976041 0.824750312331356 0.1150960216662481
### LR-특성2개 : 1573.984 0.9706807451768623 0.9775935108325122 -0.006912765655649911

### 특성 2개를 사용한 선형회귀모델의 경우 훈련정확도가 테스트정확도보다 낮게 나오는
#    -> 과소적합이 발생하고 있음. 따라서, 예측모델로 사용하기에는 부적합함
#    - 튜닝 속성이 없기에 다른 방법을 찾아야 함(다중회귀모델을 사용 -> 특성 3개 이상)
```

## 다중 회귀


```python
"""
<다중회귀모델(Multiple Regression)>
 - 여러 개의 특성(항목, 컬럼, 퓨처, 피처 모두 같은 의미)을 사용한 회귀모델
 - 특성이 많을 수록 모델은 집중력과 복잡도가 높아짐
 - 단점으로는, 훈련 시간이 오래 걸릴 수 있으며,
              시스템 성능에 따라 처리 시간이 결정됨
 - 다중회귀모델 산술식(y) = aX1 + bX2 + cX3 .... nXn + y절편
 - a, b, c ...n : 기울기(또는 계수 or 가중치라고도 칭함)
"""
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.preprocessing import StandardScaler

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
### 사용할 데이터셋 파일명 : 03_농어의_길이_높이_두께_데이터.csv
# - 읽어들인 데이터는 df 변수에 저장
# - 데이터 정보 확인하기
file_path = '../data/perch_dimensions.csv'
df  = pd.read_csv(file_path)

### 농어 무게
perch_weight = np.array(
    [5.9, 32.0, 40.0, 51.5, 70.0, 100.0, 78.0, 80.0, 85.0, 85.0,
     110.0, 115.0, 125.0, 130.0, 120.0, 120.0, 130.0, 135.0, 110.0,
     130.0, 150.0, 145.0, 150.0, 170.0, 225.0, 145.0, 188.0, 180.0,
     197.0, 218.0, 300.0, 260.0, 265.0, 250.0, 250.0, 300.0, 320.0,
     514.0, 556.0, 840.0, 685.0, 700.0, 700.0, 690.0, 900.0, 650.0,
     820.0, 850.0, 900.0, 1015.0, 820.0, 1100.0, 1000.0, 1100.0,
     1000.0, 1000.0]
     )
```

```python
"""
<농어의 길이, 두께, 너비를 이용하여 무게 예측하기>

<데이터 구분>
    - 독립변수(x) : 길이, 두께, 너비
    - 종속변수(y) : 무게
"""
print(df.shape, perch_weight.shape)
```

```python
### 데이터프레임 타입을 넘파이 배열 타입(리스트)으로 변환
# - 데이터프레임 내에 넘파이 배열로 변환하는 함수(to_numpy())
perch_full = df.to_numpy()
print(perch_full.shape, perch_weight.shape)
perch_full

train_input, test_input, train_target, test_target = train_test_split(
    perch_full, perch_weight, test_size=0.3, random_state=42
)
print(train_input.shape, train_target.shape)
print(test_input.shape, test_target.shape)
```

```python
### 모델 클래스 생성 후 오차 확인

lr = LinearRegression()
lr

#훈련 시키기(fit)
lr.fit(train_input, train_target)

# 테스트의 정확도 확인
train_score = lr.score(train_input, train_target)
test_score = lr.score(test_input, test_target)
print('정확도 체크')
print(train_score, test_score, train_score-test_score)

# 테스트 예측하기
train_pred = lr.predict(train_input)
test_pred = lr.predict(test_input)
print('테스트 예측')
print(test_pred)

# 테스트 오차 확인하기
## 평균절대오차(0에 가까울수록 좋음)
train_mae = mean_absolute_error(train_target, train_pred)
test_mae = mean_absolute_error(test_target, test_pred)
## 평균제곱오차 (0에 가까울수록 좋음)
train_mse = mean_squared_error(train_target, train_pred)
test_mse = mean_squared_error(test_target, test_pred)
## 결정계수(정확도) (1에 가까울수록 좋음)
train_r2 = r2_score(train_target, train_pred)
test_r2 = r2_score(test_target, test_pred)

print(f"훈련   - 절대오차 : {train_mae}, 제곱오차 : {train_mse},정확도 : {train_r2}")
print(f"테스트 - 절대오차 : {test_mae}, 제곱오차 : {test_mse},정확도 : {test_r2}")

```
