```python
age_median = df["Age"].median()
print(age_median)

df["Age"] = df["Age"].fillna(age_median)
df.isnull().sum()

monthlycharges_mean = df["MonthlyCharges"].mean()
print(monthlycharges_mean)

df["MonthlyCharges"].fillna(monthlycharges_mean)
```

```python
df.isnull().sum()
```

```python
import numpy as np
```

```python
# 선형 회귀는 정규 분포 가정을 따르므로,
#  - 평균으로부터 표준편차의 몇 배만큼 떨어져 있는지를 나타내는 Z-Score를 사용함
# Z = (X - μ) / σ (X: 데이터 포인트, μ: 평균, σ: 표준편차)

z_scores = np.abs((df["TotalCharges"] - df["TotalCharges"].mean()) / df["TotalCharges"].std())
print(z_scores)
# Z scores > 3 (99.7%) is considered as outlier
print("------[z-score 이상치 처리 전]-------")
df.info()
```

```python
df = df[z_scores < 3]
print("------[z-score 이상치 처리 후]-------")
df.info()
```

```python
#IQR
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)
print(f"Q1={Q1} , Q3={Q3}")

IQR= Q3-Q1
print(f"IQR= Q3-Q1={IQR}")

#Min/Max Outliers
lower_bound= Q1 - 1.5 * IQR
upper_bound= Q3 + 1.5 * IQR
print(f"lower_bound={lower_bound} , upper_bound={upper_bound}")
```

```python
print("------[IQR 이상치 처리 전]-------")
df.info()

print("------[IQR 이상치 처리 후]-------")
df = df[(df["Age"] >= lower_bound) & (df["Age"] <= upper_bound)]
df.info()
```
