### 결측치 처리

결측치를 바로 삭제하기 전에 개수와 위치를 확인하고, 빈 문자열을 `NaN`으로 통일한 다음 제거 범위를 결정한다.

#### 1. 컬럼별 결측치 개수 확인

```python
# isna()와 isnull()은 같은 기능
display(df.isna().sum())
```

#### 2. 빈 문자열을 결측값으로 변환

- 빈 문자열 `""`은 `isna()`에서 결측치로 집계되지 않음
- 먼저 빈 문자열을 `np.nan`으로 변경해야 결측치를 정확하게 확인할 수 있음

```python
# email의 빈 문자열을 NaN으로 변경
df['email'] = df['email'].replace('', np.nan)

# 변환 후 컬럼별 결측치 개수 다시 확인
display(df.isna().sum())
```

#### 3. 결측치가 있는 행 확인

```python
# 컬럼별로 결측치가 있는 행 확인
for column in ['name', 'age', 'gender', 'score', 'email']:
    missing_rows = df[df[column].isna()]
    if not missing_rows.empty:
        display(Markdown(f"**{column} 컬럼의 결측 행:**"))
        display(missing_rows)
```

#### 4. 결측치가 포함된 모든 행 제거

- `dropna()`는 하나 이상의 결측치가 있는 행을 모두 제거
- 간단하지만 데이터가 많이 줄어들 수 있으므로 제거 전후의 행 개수를 비교해야 함

```python
# 원본 df는 유지하고, 결측치가 있는 모든 행을 제거한 DataFrame 생성
df_dropna_all = df.dropna()

display(Markdown(f"**제거 전 행 개수:** {len(df)}"))
display(Markdown(f"**제거 후 행 개수:** {len(df_dropna_all)}"))
display(df_dropna_all)
```

#### 5. 특정 컬럼에 결측치가 있는 행만 제거

- `subset`을 사용하면 지정한 컬럼의 결측치만 제거 기준으로 사용
- 다른 컬럼에 결측치가 있어도 `email` 값이 있다면 해당 행은 유지됨

```python
# email이 결측치인 행만 제거
df_dropna_email = df.dropna(subset=['email'])

display(Markdown(f"**제거 전 행 개수:** {len(df)}"))
display(Markdown(f"**제거 후 행 개수:** {len(df_dropna_email)}"))
display(df_dropna_email)
```

> 분석 목적에 따라 전체 행 제거, 특정 컬럼 기준 제거, 평균·중앙값 등으로 대체하는 방법을 선택한다.

### 이상치 처리

- 이상치는 정상적인 범위를 벗어난 값
- 학생 나이의 정상 범위를 15세 이상 100세 이하로 가정
- 정상 나이만 사용해 평균을 계산한 후 나이 이상치를 평균값으로 대체

#### 1. 나이 컬럼의 이상치 확인

```python
# 정상 나이 범위 설정
normal_age_mask = df['age'].between(15, 100)

# 결측치는 제외하고 정상 범위를 벗어난 이상치만 선택
age_outlier_mask = df['age'].notna() & ~normal_age_mask

display(Markdown("**나이 이상치가 있는 행:**"))
display(df[age_outlier_mask])
```

#### 2. 정상 나이의 평균 계산

- 이상치를 평균 계산에 포함하면 평균값이 왜곡될 수 있으므로 정상 범위의 값만 사용

```python
# 결측치와 이상치를 제외한 정상 나이의 평균
mean_age = df.loc[normal_age_mask, 'age'].mean()

display(Markdown(f"**정상 나이의 평균:** {mean_age:.1f}세"))
```

#### 3. 나이 이상치를 평균값으로 대체

```python
# 원본 보호를 위한 복사본 생성
df_filled = df.copy()

# 나이 이상치를 정상 나이의 평균으로 대체
df_filled.loc[age_outlier_mask, 'age'] = round(mean_age, 1)

display(Markdown("**나이 이상치 정제 결과:**"))
display(df_filled.loc[age_outlier_mask, ['name', 'age']])
```

#### 4. 나이 결측치를 평균값으로 대체

- 이상치 처리와 결측치 처리는 서로 다른 작업이므로 단계별로 나누어 수행

```python
# 나이 결측치를 정상 나이의 평균으로 대체
df_filled['age'] = df_filled['age'].fillna(round(mean_age, 1))

display(Markdown(f"**처리 후 나이 결측치 개수:** {df_filled['age'].isna().sum()}"))
display(df_filled[['name', 'age']])
```

> 정상 범위는 데이터의 의미와 분석 목적에 따라 정해야 한다. 학생 데이터이므로 여기서는 15~100세를 예시로 사용한다.

#### 10. 결측치 및 이상치가 제거된 버전 만들기

- 원본 `df`는 유지하고 지금까지의 정제 결과를 `df_clean`에 저장
- 이름을 포함한 모든 문자열의 앞뒤 공백을 제거하고 성별 표현을 `남`과 `여`로 통일
- 빈 문자열, 결측치, 나이·점수 이상치, 잘못된 이메일 및 중복 데이터를 제거
- 마지막에 인덱스를 0부터 다시 정리

```python
# 원본 데이터를 복사해 최종 정제본 생성
df_clean = df.copy()

# 문자열 앞뒤 공백 제거 후 빈 문자열을 결측치로 통일
text_columns = df_clean.select_dtypes(include=['object', 'string']).columns
df_clean[text_columns] = df_clean[text_columns].apply(
    lambda column: column.map(lambda value: value.strip() if isinstance(value, str) else value)
)
df_clean = df_clean.replace(r'^\s*$', np.nan, regex=True)

# 성별 표현을 남/여로 통일
gender_map = {
    '남': '남', '남자': '남', '남성': '남', 'M': '남',
    '여': '여', '여자': '여', '여성': '여', 'F': '여'
}
df_clean['gender'] = df_clean['gender'].replace(gender_map)

# score를 숫자로 변환하며 변환할 수 없는 값은 결측치로 처리
df_clean['score'] = pd.to_numeric(df_clean['score'], errors='coerce')

# 이메일 형식 검사
email_pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
valid_email_mask = df_clean['email'].str.match(email_pattern, na=False)

# 결측치가 없는 행과 정상 범위의 나이·점수, 올바른 이메일만 유지
df_clean = df_clean[
    df_clean.notna().all(axis=1)
    & df_clean['age'].between(15, 100)
    & df_clean['score'].between(0, 100)
    & valid_email_mask
].copy()

# 이름과 이메일이 모두 같은 중복 행 제거
df_clean = df_clean.drop_duplicates(subset=['name', 'email'])

# 인덱스를 0부터 다시 정리
df_clean.reset_index(drop=True, inplace=True)
```

```python
# 최종 정제 결과 확인
display(Markdown(f"**원본 데이터:** {len(df)}행"))
display(Markdown(f"**최종 정제 데이터:** {len(df_clean)}행"))
display(Markdown(f"**남은 결측치:** {df_clean.isna().sum().sum()}개"))
display(Markdown(f"**남은 나이 이상치:** {(~df_clean['age'].between(15, 100)).sum()}개"))
display(Markdown(f"**남은 점수 이상치:** {(~df_clean['score'].between(0, 100)).sum()}개"))
display(Markdown(f"**남은 중복 데이터:** {df_clean.duplicated(subset=['name', 'email']).sum()}개"))
display(Markdown(f"**정리된 성별 값:** {sorted(df_clean['gender'].unique())}"))
display(df_clean)
```

- apply() 함수 내에 사용할 사용자 정의 함수 생성
- 학점 등 여러 구분으로 나눠지는 도메인 값의 컬럼을 사용할 때 유용

```python
def get_order_grade(total_price) :
    if total_price >= 80000 :
        return 'A'
    elif total_price >= 30000 :
        return 'B'
    else : 
        return 'C'

get_order_grade(90000)
```

```python
# total_price 컬럼의 값을 이용해 get_order_grade() 함수를 각 행별로 호출해서 grade 컬럼을 생성
orders['grade'] = orders['total_price'].apply(get_order_grade)
orders
```
